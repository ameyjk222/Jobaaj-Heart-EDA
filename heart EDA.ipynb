```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("heart.csv")
df
```

```python
df.info()
df.dtypes
```

```python
df.describe()
```

```python
### What is the average age of patients in the dataset ?

avg_age = df["age"].mean()
print("avg_age:", avg_age)
```

```python
### What is the gender distribution of patients ?

gender_distribution = df["sex"].value_counts()
print("gender_distribution:", gender_distribution)

gender_distribution = df["sex"].value_counts(normalize=True) * 100
print(gender_distribution)
```

```python
### What is the average resting blood pressure of patients ?

resting_bp = df["trestbps"].mean()
print ("resting_bp:", resting_bp)
```

```python
### How many patients have fasting blood sugar levels higher than 120 mg/dl?
fasting_blood_sugar = df["fbs"].sum()
print("fasting_blood_sugar:", fasting_blood_sugar)
```

```python
### What are the different types of chest pain recorded in the dataset ?
types_of_chest_pain = df["cp"].unique()
print("types_of_chest_pain:", types_of_chest_pain)
```

```python
### What is the maximum heart rate achieved by patients ?
max_heart_rate = df["thalach"].max()
print("max_heart_rate:", max_heart_rate)
```

```python
### What percentage of patients experience exercise-induced angina ?
patients_percentage = df["exang"].mean() * 100 
print("patients_percentage:", patients_percentage)
```

```python
### What is the average cholesterol level in the dataset ?
avg_chol = df["chol"].mean()
print("avg_chol:", avg_chol)
```

```python
### How many patients have a resting electrocardiographic result of 2 ?
resting_etc = df ["restecg"].value_counts()[2]
print("resting_etc:", resting_etc)
```

```python
### What is the distribution of the number of major vessels colored by fluoroscopy ?
fluroscopy = df ["ca"].value_counts()
print("fluroscopy:", fluroscopy)
```

```python
### What is the correlation between age and cholestrol levels ?

corr = df ["age"]. corr(df["chol"])
print("corr:", corr)
```

```python
### What is the distribution of chest pain types across different age groups ?
distribution_pain = df.groupby('age')['cp'].value_counts().unstack().plot(
    kind="bar",
    stacked=True,
    figsize=(18,6)
)

plt.title("Distribution of chest pain by Across different age groups")
plt.xlabel("Age")
plt.ylabel("Chest Pain")
plt.legend(title='Chest Pain Type')
plt.show()
```

```python
### How does maximum heart rate vary with exercise_induced angina ?

maximum_heart = df.groupby ("exang")["thalach"].mean()
print("maximum_heart:", maximum_heart)
```

```python
### Is there a significant difference in resting blood pressure between male and female patients ?

diff_blood_pressure = df.groupby ("sex")["trestbps"].mean()
print("diff_blood_pressure:", diff_blood_pressure)
```

```python
### What is the relationship between fasting blood sugar levels and the presence of heart disease ?

fasting_bp = pd.crosstab(['fbs'], df['target']).plot(
    kind="bar",
    stacked=True,
    figsize = (16,9)
)

plt.title("difference in resting blood pressure for Male and Female")
plt.xlabel("Fbs")
plt.ylabel("target")
plt.legend("Fasting Blood Sugar")
plt.show()
```

```python
### How does the number of major vessels (ca) affect the target variable (heart disease presence)?

pd.crosstab(df['ca'], df['target'])

target_variable = pd.crosstab (df['ca'], df['target']).plot(
    kind="bar",
    stacked=True,
    figsize=(8,5)
)

plt.title("Number of Major Vessels affect the Target variable")
plt.xlabel("ca")
plt.ylabel("target")
plt.legend("Heart Disease affects")
plt.show()
```

```python
### What is the average oldpeak value for patients with different types of chest pain ?

avg_oldpeak = df.groupby ("cp")["oldpeak"].mean()
print("avg_oldpeak:", avg_oldpeak)
```

```python
### Analyze the distribution of thalassemia types (thal) among patients with heart disease.

thal_distribution = pd.crosstab(df['thal'], df['target']).plot(
    kind="bar",
    stacked=True,
    figsize=(8,6)
)

plt.title("Thalassemia types with Heart Disease")
plt.xlabel("Thalassemia")
plt.xlabel("Target")
plt.legend("Thalassemia")
plt.show()
```

```python
### What are the most common combinations of risk factors in patients with heart disease ?

risk_combination = (
    df[df['target'] == 1].groupby(['cp', 'fbs', 'exang', 'thal'])
    .size()
    .reset_index(name='counts')
    .sort_values(by='counts', ascending=False))

print(risk_combination.head(10))
```

```python
### Perform a pairwise comparison of clinical measurements for patients with and without heart disease.

## Patients with heart disease
heart_disease = df[df['target'] == 1]

heart_disease.describe()


## Patients without heart disease
no_heart_disease = df[df['target'] == 0]

no_heart_disease.describe()

comparison = pd.DataFrame({
    'heart_disease' :df[df['target'] == 1].mean(),
    'no_heart_disease' :df[df['target'] == 0].mean()
})

print(comparison)
```

```python
### What is the effect of combining multiple risk factors (age, cholestrol, blood pressure) on the likelihood of heart disease ?

import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot (
    df,
    hue='target',
    vars = ['age', 'chol', 'trestbps']
)

plt.show()
```

```python
### Which clinical measurement has the strongest correlation with heart disease presence ?

correlation_with_target = df.corr(numeric_only=True)['target'].sort_values(ascending=False)

print(correlation_with_target)

correlation_with_target.plot (

    kind="bar",
    figsize = (8,6)
)

plt.title("Correlation with Heart Disease")
plt.ylabel("Correlation")
plt.show()
```
